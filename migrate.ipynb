```python
# 1) Move os CSVs para as novas pastas
moves = {
    "base_atendentes.csv":      "atendentes",
    "canais.csv":               "canais",
    "chamados.csv":             "chamados",
    "Chamados_Hora.CSV":        "chamados_hora",
    "clientes.csv":             "clientes",
    "custos.csv":               "custos",
    "base_motivos.csv":         "motivos",
    "pesquisa_satisfacao.csv":  "pesquisas_de_satisfacao",
}
for fname, folder in moves.items():
    dbutils.fs.mkdirs(f"{landing_path}/{folder}")
    dbutils.fs.mv(f"{landing_path}/{fname}", f"{landing_path}/{folder}/{fname}")

# 2) Zera tabelas Bronze e checkpoints
tables = ["tb_atendentes", "tb_canais", "tb_chamados", "tb_chamados_hora",
          "tb_clientes", "tb_custos", "tb_motivos", "tb_pesquisas_de_satisfacao"]
for t in tables:
    spark.sql(f"DROP TABLE IF EXISTS {bronze_schema}.{t}")
    dbutils.fs.rm(f"{checkpoint_path}/{t}", recurse=True)

# Schemas antigos (nomes por arquivo e o "default")
dbutils.fs.rm(f"{checkpoint_path}/schemas", recurse=True)
```
